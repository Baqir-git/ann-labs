## Exercise 13 — Vectorized Neuron

Consider:

\[

\]


Use NumPy to calculate the neuron output.


````python
x = np.array([1, 2, 3])
w = np.array([0.2, 0.5, 0.8])
b = 1

z = np.dot(w, x) + b

print("Neuron output:", z)
````

## Exercise 17 — Step Activation Function

A neuron usually applies an **activation function** after computing the weighted sum.

For this first example, use a step function:

\[
f(z)=
\begin{cases}
1 & z\geq0\\
0 & z<0
\end{cases}
\]

Implement this function in Python.


````python
def step_function(z):
    if z >= 0:
        return 1
    else:
        return 0

print(step_function(4.4))
print(step_function(-2))
````

## Complete Artificial Neuron

Now combine:

1. Inputs
2. Weights
3. Bias
4. Weighted sum
5. Activation

into one function.


````python
def neuron_predict(x1, x2, w1, w2, b):
    # Weighted sum
    z = w1 * x1 + w2 * x2 + b

    # Step activation
    if z >= 0:
        output = 1
    else:
        output = 0

    return output


output = neuron_predict(
    x1=2,
    x2=3,
    w1=0.5,
    w2=0.8,
    b=1
)

print("Prediction:", output)
````

# Final Challenge
### 115–120 minutes

## Problem

Create a function called `predict()` that:

1. receives two inputs
2. receives two weights
3. receives a bias
4. computes

\[
z=w_1x_1+w_2x_2+b
\]

5. applies the step activation
6. returns either `0` or `1`

### Test Case

```python
x1 = 4
x2 = -2
w1 = 0.5
w2 = 0.7
b = -1
```

### Before looking at the solution

Calculate the result manually.


````python
# Try to solve this yourself first.

def predict(x1, x2, w1, w2, b):
    # Write your solution here
    pass
````

## Solution


````python
def predict(x1, x2, w1, w2, b):
    z = w1 * x1 + w2 * x2 + b

    if z >= 0:
        return 1
    else:
        return 0


result = predict(
    x1=4,
    x2=-2,
    w1=0.5,
    w2=0.7,
    b=-1
)

print("Prediction:", result)
````
